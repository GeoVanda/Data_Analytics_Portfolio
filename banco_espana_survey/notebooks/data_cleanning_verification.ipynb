```python
"""
DATA CLEANING AND VERIFICATION
Financial Competencies Survey - Bank of Spain (2016 & 2021)

This notebook loads and cleans the raw data using the ETL pipeline
"""

import importlib.util
from pathlib import Path
import pandas as pd
import numpy as np

# Load ETL pipeline module
module_path = Path.cwd() / "scripts" / "pipeline_etl_ecf.py"
spec = importlib.util.spec_from_file_location("pipeline_etl_ecf", str(module_path))
pipeline_etl_ecf = importlib.util.module_from_spec(spec)
spec.loader.exec_module(pipeline_etl_ecf)

# 1. LOAD AND CLEAN DATA
df_2016 = pipeline_etl_ecf.load_clean_transform("data/ecf_2016.csv", 2016)
df_2021 = pipeline_etl_ecf.load_clean_transform("data/ecf_2021.csv", 2021)

# Concatenate both years
df_all = pd.concat([df_2016, df_2021], ignore_index=True)
print(f"\nCombined dataset: {df_all.shape}")

# 2. VERIFY DATA QUALITY

# Check for missing values
print("\nMissing values per column:")

print(df_all.isna().sum())
print(f"\nTotal missing values: {df_all.isna().sum().sum()}")

# Check data types
print("\nData types:")
print(df_all.dtypes)

# Check unique values in categorical columns
print("\nUnique values in categorical columns:")
print(f"Years: {df_all['year'].unique()}")
print(f"Age brackets: {sorted(df_all['age_bracket'].unique())}")
print(f"Education levels: {sorted(df_all['edu_level'].unique())}")
print(f"Labor categories: {sorted(df_all['labor_categ'].unique())}")
```
